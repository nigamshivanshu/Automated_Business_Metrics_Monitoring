## Step 8: KPI Computation Design

---

### KPI Computation Principles

- KPIs are computed only on validated data
- Each KPI has a single, documented definition
- Grain alignment is mandatory
- Growth metrics require historical context

---

### KPI Definitions

#### Revenue
Total monetary value of completed orders.

Formula:
Revenue = SUM(order_amount)

---

#### Orders
Total number of completed orders.

Formula:
Orders = COUNT(DISTINCT order_id)

---

#### Average Order Value (AOV)
Average revenue per order.

Formula:
AOV = Revenue / Orders

If Orders = 0, AOV is set to NULL.

---

#### Week-over-Week (WoW) Growth
Percentage change compared to the previous week.

Formula:
WoW % = (Current Week - Previous Week) / Previous Week * 100

Computed only when prior period exists.

---

#### Month-over-Month (MoM) Growth
Percentage change compared to the previous month.

Formula:
MoM % = (Current Month - Previous Month) / Previous Month * 100

---

### Dimensional Breakdowns

KPIs are computed by:
- Date
- Platform
- Location

---

### KPI Validation

- Cross-check daily vs weekly rollups
- Identify unexplained spikes or drops
- Validate totals against raw data
- Surface data quality flags alongside KPIs

---

### KPI Output Structure

Each KPI output includes:
- Time period

- KPI values

- Data quality indicators

```python
# Database and Data Generation

import pandas as pd
import numpy as np
import sqlite3
from datetime import datetime, timedelta
import random
```

```python
import os

os.makedirs("data", exist_ok=True)

db_path = "data/business_metrics.db"
conn = sqlite3.connect(db_path)

conn
```

```python
#Creating Dimension Table

#Platforms

dim_platforms = pd.DataFrame({
    "platform_id": [1, 2, 3],
    "platform_name": ["Web", "Mobile App", "Marketplace"]
})

dim_platforms

# Locations

dim_locations = pd.DataFrame({
    "location_id": [1, 2, 3, 4],
    "location_name": ["Delhi", "Mumbai", "Bangalore", "Hyderabad"],
    "region": ["North", "West", "South", "South"]
})

dim_locations

# Customers

num_customers = 2000

dim_customers = pd.DataFrame({
    "customer_id": range(1, num_customers + 1),
    "customer_segment": np.random.choice(
        ["New", "Returning", "VIP"],
        size=num_customers,
        p=[0.5, 0.35, 0.15]
    ),
    "signup_date": pd.to_datetime("2024-01-01") +
                   pd.to_timedelta(
                       np.random.randint(0, 730, size=num_customers),
                       unit="D"
                   )
})


dim_customers.head()

# Dates

start_date = datetime(2024, 1, 1)
end_date = datetime(2025, 12, 31)  # full 2025 coverage

dates = pd.date_range(start=start_date, end=end_date, freq="D")

dim_dates = pd.DataFrame({
    "date_id": range(1, len(dates) + 1),
    "date": dates,
    "day": dates.day,
    "week": dates.isocalendar().week,
    "month": dates.month,
    "year": dates.year,
    "is_weekend": dates.weekday >= 5
})

dim_dates.head()
```

```python
# Create Fact Table — Orders

num_orders = 50000

fact_orders = pd.DataFrame({
    "order_id": range(1, num_orders + 1),
    "order_date": np.random.choice(dim_dates["date"], size=num_orders),
    "customer_id": np.random.choice(dim_customers["customer_id"], size=num_orders),
    "platform_id": np.random.choice(dim_platforms["platform_id"], size=num_orders, p=[0.5, 0.3, 0.2]),
    "location_id": np.random.choice(dim_locations["location_id"], size=num_orders),
    "order_amount": np.round(np.random.gamma(shape=2.0, scale=800, size=num_orders), 2),
    "order_status": np.random.choice(
        ["completed", "cancelled", "returned"],
        size=num_orders,
        p=[0.9, 0.07, 0.03]
    ),
    "created_at": pd.Timestamp.now()
})

fact_orders.head()

```

```python
recent_dates = dim_dates["date"].sample(
    frac=1,
    weights=np.linspace(0.2, 1.0, len(dim_dates))
)

fact_orders["order_date"] = np.random.choice(
    recent_dates,
    size=num_orders
)
```

```python
# Insert Data into SQLite

dim_platforms.to_sql("dim_platforms", conn, if_exists="replace", index=False)
dim_locations.to_sql("dim_locations", conn, if_exists="replace", index=False)
dim_customers.to_sql("dim_customers", conn, if_exists="replace", index=False)
dim_dates.to_sql("dim_dates", conn, if_exists="replace", index=False)
fact_orders.to_sql("fact_orders", conn, if_exists="replace", index=False)
```

```python
tables = ["fact_orders", "dim_customers", "dim_platforms", "dim_locations", "dim_dates"]

for table in tables:
    count = pd.read_sql(f"SELECT COUNT(*) AS cnt FROM {table}", conn)
    print(f"{table}: {count.iloc[0]['cnt']}")
```

```python
# Creating Folder named SQL to store sql files

import os
os.makedirs("sql", exist_ok=True)
```

```python
base_orders_sql = """
SELECT
    fo.order_id,
    fo.order_date,
    d.year,
    d.month,
    d.week,
    fo.customer_id,
    fo.platform_id,
    p.platform_name,
    fo.location_id,
    l.location_name,
    l.region,
    fo.order_amount
FROM fact_orders fo
JOIN dim_dates d
    ON fo.order_date = d.date
JOIN dim_platforms p
    ON fo.platform_id = p.platform_id
JOIN dim_locations l
    ON fo.location_id = l.location_id
WHERE fo.order_status = 'completed'
  AND fo.order_amount > 0
  AND fo.order_date <= DATE('now');
"""

with open("sql/base_orders_extract.sql", "w") as f:
    f.write(base_orders_sql)
```

```python
daily_kpi_sql = """
SELECT
    d.date AS metric_date,
    d.year,
    d.month,
    fo.platform_id,
    p.platform_name,
    fo.location_id,
    l.location_name,
    COUNT(DISTINCT fo.order_id) AS orders,
    SUM(fo.order_amount) AS revenue,
    ROUND(SUM(fo.order_amount) / COUNT(DISTINCT fo.order_id), 2) AS aov
FROM fact_orders fo
JOIN dim_dates d
    ON fo.order_date = d.date
JOIN dim_platforms p
    ON fo.platform_id = p.platform_id
JOIN dim_locations l
    ON fo.location_id = l.location_id
WHERE fo.order_status = 'completed'
  AND fo.order_amount > 0
GROUP BY
    d.date,
    d.year,
    d.month,
    fo.platform_id,
    p.platform_name,
    fo.location_id,
    l.location_name;
"""

with open("sql/daily_kpi_aggregation.sql", "w") as f:
    f.write(daily_kpi_sql)
```

```python
weekly_kpi_sql = """
SELECT
    d.year,
    d.week,
    fo.platform_id,
    p.platform_name,
    fo.location_id,
    l.location_name,
    COUNT(DISTINCT fo.order_id) AS orders,
    SUM(fo.order_amount) AS revenue,
    ROUND(SUM(fo.order_amount) / COUNT(DISTINCT fo.order_id), 2) AS aov
FROM fact_orders fo
JOIN dim_dates d
    ON fo.order_date = d.date
JOIN dim_platforms p
    ON fo.platform_id = p.platform_id
JOIN dim_locations l
    ON fo.location_id = l.location_id
WHERE fo.order_status = 'completed'
  AND fo.order_amount > 0
GROUP BY
    d.year,
    d.week,
    fo.platform_id,
    p.platform_name,
    fo.location_id,
    l.location_name;
"""

with open("sql/weekly_kpi_aggregation.sql", "w") as f:
    f.write(weekly_kpi_sql)
```

```python
# Standardize Column Names

base_orders.columns = (
    base_orders.columns
    .str.strip()
    .str.lower()
    .str.replace(" ", "_")
)
```

```python
# Enforce Data Types

base_orders["order_date"] = pd.to_datetime(base_orders["order_date"])
base_orders["order_amount"] = pd.to_numeric(base_orders["order_amount"])
```

```python
# Remove Exact Duplicates (if any)

before = len(base_orders)

base_orders = base_orders.drop_duplicates(subset=["order_id"])

after = len(base_orders)

print(f"Removed {before - after} duplicate orders")
```

```python
# Row-Level Validation Flags

base_orders["is_negative_revenue"] = base_orders["order_amount"] <= 0
base_orders["is_future_date"] = base_orders["order_date"] > pd.Timestamp.today()
base_orders["is_missing_key"] = (
    base_orders[["order_id", "platform_id", "location_id"]].isnull().any(axis=1)
)
```

```python
# Uniqueness Check
base_orders["is_duplicate_order"] = base_orders.duplicated(subset=["order_id"], keep=False)
```

```python
# Aggregate-Level Validation (Business-Level)
```

```python
# Daily Revenue Calculation (for anomaly detection)

daily_revenue = (
    base_orders
    .groupby(base_orders["order_date"].dt.date)["order_amount"]
    .sum()
    .reset_index(name="daily_revenue")
)
```

```python
# Detect Abnormal Revenue Spikes/Drops

daily_revenue["pct_change"] = daily_revenue["daily_revenue"].pct_change()

daily_revenue["is_anomalous_day"] = daily_revenue["pct_change"].abs() > 0.5
```

```python
# Keep only ONE order_date column
base_orders = base_orders.loc[:, ~base_orders.columns.duplicated()]

# If order_date_x exists, normalize
if "order_date_x" in base_orders.columns:
    base_orders.rename(columns={"order_date_x": "order_date"}, inplace=True)

# Drop any lingering order_date_y
base_orders.drop(columns=[c for c in ["order_date_y"] if c in base_orders.columns], inplace=True)
```

```python
# Remove duplicate columns by keeping the first occurrence
base_orders = base_orders.loc[:, ~base_orders.columns.duplicated()]
```

```python
base_orders["order_date"] = pd.to_datetime(base_orders["order_date"])
base_orders["order_date_key"] = base_orders["order_date"].dt.date
```

```python
base_orders = base_orders.merge(
    daily_revenue[["order_date", "is_anomalous_day"]],
    left_on="order_date_key",
    right_on="order_date",
    how="left"
)


base_orders.drop(columns=["order_date_key", "order_date_y"], inplace=True)
base_orders.rename(columns={"order_date_x": "order_date"}, inplace=True)
```

```python
required_columns = [
    "is_negative_revenue",
    "is_future_date",
    "is_duplicate_order",
    "is_anomalous_day"
]

missing = [c for c in required_columns if c not in base_orders.columns]
missing
```

```python
# Recomputing to check for is_anomalous_day column

daily_revenue = (
    base_orders
    .assign(order_date_day=base_orders["order_date"].dt.date)
    .groupby("order_date_day", as_index=False)["order_amount"]
    .sum()
    .rename(columns={"order_amount": "daily_revenue"})
)
```

```python
# Detect Daily Anomalies
daily_revenue["pct_change"] = daily_revenue["daily_revenue"].pct_change()

daily_revenue["is_anomalous_day"] = daily_revenue["pct_change"].abs() > 0.5
```

```python
#Merge Anomaly Flag Back (NO COLLISIONS)
base_orders = (
    base_orders
    .assign(order_date_day=base_orders["order_date"].dt.date)
    .merge(
        daily_revenue[["order_date_day", "is_anomalous_day"]],
        on="order_date_day",
        how="left"
    )
    .drop(columns=["order_date_day"])
)
```

```python
# Data Quality Summary

data_quality_summary = {
    "total_orders": len(base_orders),
    "negative_revenue_orders": int(base_orders["is_negative_revenue"].sum()),
    "future_dated_orders": int(base_orders["is_future_date"].sum()),
    "duplicate_orders": int(base_orders["is_duplicate_order"].sum()),
    "anomalous_days": int(base_orders["is_anomalous_day"].sum())
}

pd.DataFrame.from_dict(data_quality_summary, orient="index", columns=["count"])
```
